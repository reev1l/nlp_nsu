```python
import re
import nltk
import pandas as pd
from pymystem3 import Mystem
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer
```

```python
nltk.download('stopwords')
data = pd.read_csv('tweets.csv')
m = Mystem()

def lemmatize(text):
    return ''.join(m.lemmatize(text))

def clear_text(text):
    text = re.sub(r'[^а-яА-ЯёЁ ]', ' ', text)
    return ' '.join(text.split())
```

```python
lemmatized = {
    text: clear_text(lemmatize(text))
    for text in data['text'].unique()
}
data['lemm_text'] = data['text'].map(lemmatized)
data.head()
```

```python
corpus = data['lemm_text'].values.astype('U')
stop = stopwords.words('russian')
print(stop)
```

```python
count_with_stopwords = CountVectorizer()
bow_with_stopwords = count_with_stopwords.fit_transform(corpus)

count_without_stopwords = CountVectorizer(stop_words=stop)
bow_without_stopwords = count_without_stopwords.fit_transform(corpus)

print('Размер BoW со стоп-словами:', bow_with_stopwords.shape)
print('Размер BoW без стоп-слов:', bow_without_stopwords.shape)
```

```python
count_bigrams = CountVectorizer(ngram_range=(2, 2))
bigrams = count_bigrams.fit_transform(corpus)

print('Размер матрицы биграмм:', bigrams.shape)
```
