```python
import pandas as pd
from pymystem3 import Mystem

data = pd.read_csv('tweets.csv')
corpus = data['text'].values.astype('U')
m = Mystem()
```

```python
def lemmatize(text):
    return ''.join(m.lemmatize(text))
```

```python
first_text = corpus[0]
print(first_text)
print(lemmatize(first_text))
```
