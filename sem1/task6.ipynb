```python
from math import log10

corpus = [
    'Карл у Клара украсть коралл',
    'Клара у Карл украсть кларнет',
    'Клара строго карать Карл за кража коралл'
]
word = 'коралл'
text = corpus[2]
```

```python
words = text.lower().split()
t = words.count(word)
n = len(words)
D = len(corpus)
d = sum(word in document.lower().split() for document in corpus)

tf = t / n
idf = log10(D / d)
tf_idf = tf * idf

print('TF:', tf)
print('IDF:', idf)
print('TF-IDF:', tf_idf)
```
